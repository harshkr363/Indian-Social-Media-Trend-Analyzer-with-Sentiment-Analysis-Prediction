```python
import pandas as pd

df_full = pd.read_csv(r"Dataset\india_subreddit_comments.csv")
df = df_full.sample(n=60000, random_state=42)
```

```python
#Check column
df.columns
```

```python
#Understanding Data
df.head(10)
```

```python
#Creating text column
df["text"] = df["body"]
```

```python
#Keeping only needed column
df = df[["text", "score"]]
df.dropna(inplace=True)
```

```python
#Text Cleaning
import re

def clean_text(text):
    text = str(text)
    text = re.sub(r"http\S+", "", text)
    text = re.sub(r"@\w+", "", text)
    text = re.sub(r"[^A-Za-z\s]", "", text)
    text = text.lower()
    return text

df["clean_text"] = df["text"].apply(clean_text)
```

```python
#Creating labels for further sentiment anlysis
def get_sentiment(score):
    if score > 2:
        return "positive"
    elif score < 0:
        return "negative"
    else:
        return "neutral"

df["label"] = df["score"].apply(get_sentiment)
```

```python
#Selecting Only Final Column
df = df[["text", "clean_text", "label"]]
```

```python
#Adding Metadata
df["source"] = "reddit_india"
df["topic"] = "general"
```

```python
#checking Cleaned Dataset
print(df.head())
print(df["label"].value_counts())
print(df.shape)
```

```python
#Saving Cleaned Dataset
df.to_csv("indian_dataset_clean.csv", index=False)
```
